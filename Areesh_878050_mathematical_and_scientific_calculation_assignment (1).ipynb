7) Compound Interest

The PDF gives the compound-interest formula and defines P, r, n, and t.

```python
P = float(input("Enter principal amount: "))
r = float(input("Enter annual interest rate (decimal): "))
n = int(input("Enter number of times compounded per year: "))
t = float(input("Enter time in years: "))

A = P * (1 + r / n) ** (n * t)

compound_interest = A - P

print("Total Amount =", A)
print("Compound Interest =", compound_interest)
```

11) Quadratic Equation Roots

Formula: x = (-b ± √(b² - 4ac)) / 2a

```python
import cmath

a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

discriminant = b ** 2 - 4 * a * c

x1 = (-b + cmath.sqrt(discriminant)) / (2 * a)
x2 = (-b - cmath.sqrt(discriminant)) / (2 * a)

print("Root 1 =", x1)
print("Root 2 =", x2)
```

13) Gravitational Force

Formula: F = G(m1 × m2) / r²

```python
G = 6.67430e-11

m1 = float(input("Enter mass of object 1 in kg: "))
m2 = float(input("Enter mass of object 2 in kg: "))
r = float(input("Enter distance between centers in meters: "))

force = G * m1 * m2 / (r ** 2)

print("Gravitational Force =", force, "N")
```

18) Future Value in Savings

19) Work Done by a Force

Formula: W = f × d × cos(θ)

```python
import math

f = float(input("Enter force: "))
d = float(input("Enter distance: "))
theta = float(input("Enter angle in degrees: "))

work = f * d * math.cos(math.radians(theta))

print("Work Done =", work, "J")
```

20) Heat Transfer

Formula: Q = mc(ΔT)

```python
m = float(input("Enter mass: "))
c = float(input("Enter specific heat capacity: "))
delta_T = float(input("Enter change in temperature: "))

Q = m * c * delta_T

print("Heat Transferred =", Q)
```
